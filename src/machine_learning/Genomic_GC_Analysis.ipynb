```python
# Our first cloud bioinformatics calculation
dna_sequence = "ATCGATCGATGGATCGATCG"

# Calculate total length
total_length = len(dna_sequence)

# Calculate GC count using built-in counts
g_count = dna_sequence.count("G")
c_count = dna_sequence.count("C")
gc_total = g_count + c_count

# Compute percentage matrix score
gc_percentage = (gc_total / total_length) * 100

print("🧬 --- Genomic Sequence Analysis --- 🧬")
print(f"Sequence: {dna_sequence}")
print(f"Total Base Pairs: {total_length}")
print(f"GC Percentage Score: {gc_percentage:.2f}%")
```
